```python

# 1. Import libraries
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Include required forrest classifier lib
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_classification

from google.colab import files
uploaded = files.upload()

# 3. Load the dataset
data = pd.read_csv("breast-cancer.csv")

print("First five rows:")
print(data.head())


# 3. Define features and target
X = data.drop(columns=["id", "diagnosis"]) # -> Dropped id and diagnosis columns in X

y = data["diagnosis"] # -> Assign Diagnosis


# Split the data into training and testing sets
# https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state=42)

# X: the input data the model learns from
# y: the labels we are trying to predict [M, B]
# test_size = 0.20: 20% of the rows go into the test set, this means the remaining 80% go into the training set
# random_state = 42: This is the RNG seed -> this is to show that the results are reproducible using the exact same split

# X_train: receives 80% of the data to train the model
# X_test: receives 20% of the data to test the model
# y_train: labels attached to the features of the training set
# y_test: same as above but for the testing set


# Scale the features
# What is feature scaling?
# Feature scaling forces all features/attributes to be calculated/weighed as of equal importance

# Standardization scaling: Zscore = (x - Mean) / (Standard Deviation)
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)



# Create the k-NN classifier
knn = KNeighborsClassifier(n_neighbors = 5) # 5 nearest neighbours

#TODO
# Check variation of nerest neighbours: 3, 5, 7, 11

# Train the model
knn.fit(X_train_scaled, y_train)

# Make predictions
y_pred = knn.predict(X_test_scaled)


# KNN
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    pos_label="M"
)

recall = recall_score(
    y_test,
    y_pred,
    pos_label="M"
)

f1 = f1_score(
    y_test,
    y_pred,
    pos_label="M"
)

print("\nKNN Results")
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-Score :", f1)


# === Random Forest ==== #

# RF classifier
rf = RandomForestClassifier(n_estimators = 2, random_state = 42)

# RF Build forest of trees
rf.fit(X_train, y_train)

# RF Predict
rf_pred = rf.predict(X_test)

# RF Results
rf_accuracy = accuracy_score(y_test, rf_pred)

rf_precision = precision_score(
    y_test,
    rf_pred,
    pos_label="M"
)

rf_recall = recall_score(
    y_test,
    rf_pred,
    pos_label="M"
)

rf_f1 = f1_score(
    y_test,
    rf_pred,
    pos_label="M"
)



print("\nRF Results")
print("Accuracy: ", rf_accuracy)
print("Precision:", rf_precision)
print("Recall   :", rf_recall)
print("F1-Score :", rf_f1)


# Part A)
# Questions C) Why is false negatives particularly mportant w hen interpreting results?
#


```
